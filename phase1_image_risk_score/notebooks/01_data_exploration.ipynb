## 目的
- DTU原データのフォルダ構成・画像数・解像度を把握する
- 損傷クラスの分布を確認する
- 2017年・2018年の時点差を確認する

## データ
- **種別**: 一次ソース
- **出典**: DTU / Mendeley (DOI: 10.17632/hd96prn3nc.2)
- **保存先**: `../data/raw/`

```python
from pathlib import Path
import os

# リポジトリルートからの相対パスで定義
REPO_ROOT = Path("../../")  # notebooksフォルダから2階層上
DATA_RAW = REPO_ROOT / "data" / "raw"

print(f"データパス: {DATA_RAW.resolve()}")
print(f"存在確認: {DATA_RAW.exists()}")
```

```python
# フォルダ構成の確認
for item in sorted(DATA_RAW.rglob("*")):
    print(item.relative_to(DATA_RAW))
```

```python
import matplotlib.pyplot as plt
from PIL import Image
import numpy as np

# 画像ファイル一覧を取得
image_files = list(DATA_RAW.rglob("*.jpg")) + list(DATA_RAW.rglob("*.png"))
print(f"総画像数: {len(image_files)}")

# 年別の枚数
for year in ["2017", "2018"]:
    count = sum(1 for f in image_files if year in str(f))
    print(f"{year}年: {count}枚")
```

```python
# 解像度の確認（サンプル10枚）
resolutions = []
for f in image_files[:10]:
    img = Image.open(f)
    resolutions.append(img.size)
    print(f"{f.name}: {img.size}")
```

```python
# サンプル画像の表示
fig, axes = plt.subplots(2, 3, figsize=(15, 8))
sample_files = image_files[:6]

for ax, f in zip(axes.flat, sample_files):
    img = Image.open(f)
    ax.imshow(img)
    ax.set_title(f.name, fontsize=8)
    ax.axis("off")

plt.tight_layout()
plt.savefig("../reports/sample_images.png", dpi=100)
plt.show()
```
